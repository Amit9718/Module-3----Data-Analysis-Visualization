## 1. Introduction
This notebook performs Exploratory Data Analysis (EDA) on the Titanic dataset to uncover patterns related to passenger survival, demographics, and fare distribution.

## 2. Dataset Description
The Titanic dataset contains information about 891 passengers aboard the RMS Titanic. Key columns:
- `survived` — 0 = No, 1 = Yes
- `pclass` — Passenger class (1st, 2nd, 3rd)
- `sex` — Gender
- `age` — Age in years
- `fare` — Ticket fare
- `embarked` — Port of embarkation (C, Q, S)

## 3. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['figure.figsize'] = (10, 5)
sns.set_theme(style='whitegrid')
print('Libraries loaded successfully.')
```

## 4. Load Dataset

```python
df = sns.load_dataset('titanic')
print(f'Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns')
```

## 5. Data Exploration

```python
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nData types:\n', df.dtypes)
df.head()
```

```python
print('Descriptive Statistics:')
df.describe()
```

```python
print('Missing values per column:')
print(df.isnull().sum())
```

## 6. Data Cleaning

```python
# Fill missing age with median
df['age'].fillna(df['age'].median(), inplace=True)

# Fill missing embarked with mode
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)

# Drop 'deck' column — too many missing values
df.drop(columns=['deck'], inplace=True)

# Drop duplicates
before = len(df)
df.drop_duplicates(inplace=True)
print(f'Duplicates removed: {before - len(df)}')

print('Remaining missing values:')
print(df.isnull().sum())
```

## 7. Data Analysis

```python
# Overall survival rate
survival_rate = df['survived'].mean() * 100
print(f'Overall survival rate: {survival_rate:.1f}%')

# Survival by gender
print('\nSurvival rate by gender:')
print(df.groupby('sex')['survived'].mean().apply(lambda x: f'{x*100:.1f}%'))

# Survival by class
print('\nSurvival rate by class:')
print(df.groupby('pclass')['survived'].mean().apply(lambda x: f'{x*100:.1f}%'))

# Average fare by class
print('\nAverage fare by class:')
print(df.groupby('pclass')['fare'].mean().round(2))

# Age statistics
print(f'\nAge — Min: {df["age"].min()}, Max: {df["age"].max()}, Mean: {df["age"].mean():.1f}')
```

```python
# Correlation matrix
numeric_cols = df.select_dtypes(include=np.number)
print('Correlation matrix (key columns):')
numeric_cols[['survived','pclass','age','fare','sibsp','parch']].corr()
```

## 8. Visualizations

```python
# Bar Chart — Survival count by passenger class
fig, ax = plt.subplots()
survival_by_class = df.groupby('pclass')['survived'].sum()
ax.bar(['1st Class', '2nd Class', '3rd Class'], survival_by_class.values,
       color=['#2ecc71', '#3498db', '#e74c3c'])
ax.set_title('Survivors by Passenger Class')
ax.set_xlabel('Passenger Class')
ax.set_ylabel('Number of Survivors')
for i, v in enumerate(survival_by_class.values):
    ax.text(i, v + 1, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Line Chart — Average fare by age group
df['age_group'] = pd.cut(df['age'], bins=[0,12,18,35,60,100],
                          labels=['Child','Teen','Young Adult','Adult','Senior'])
avg_fare = df.groupby('age_group', observed=True)['fare'].mean()

fig, ax = plt.subplots()
ax.plot(avg_fare.index.astype(str), avg_fare.values, marker='o', color='#3498db', linewidth=2)
ax.set_title('Average Fare by Age Group')
ax.set_xlabel('Age Group')
ax.set_ylabel('Average Fare ($)')
plt.tight_layout()
plt.show()
```

```python
# Pie Chart — Embarkation port distribution
port_counts = df['embarked'].value_counts()
port_labels = {'S': 'Southampton', 'C': 'Cherbourg', 'Q': 'Queenstown'}

fig, ax = plt.subplots()
ax.pie(port_counts.values,
       labels=[port_labels.get(p, p) for p in port_counts.index],
       autopct='%1.1f%%', startangle=140,
       colors=['#3498db', '#e74c3c', '#2ecc71'])
ax.set_title('Passenger Embarkation Ports')
plt.tight_layout()
plt.show()
```

```python
# Grouped bar — Survival rate by gender and class
pivot = df.groupby(['pclass','sex'])['survived'].mean().unstack() * 100
pivot.plot(kind='bar', color=['#e74c3c','#3498db'], edgecolor='black')
plt.title('Survival Rate by Class and Gender (%)')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.xticks(ticks=[0,1,2], labels=['1st','2nd','3rd'], rotation=0)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

## 9. Key Findings

1. **Overall survival rate was ~38%** — less than half of passengers survived.
2. **Women survived at a much higher rate (~74%)** compared to men (~19%), consistent with 'women and children first' protocol.
3. **1st class passengers had the highest survival rate (~63%)**, followed by 2nd class (~47%) and 3rd class (~24%).
4. **1st class passengers paid significantly higher fares** (avg ~$84) vs 3rd class (~$13).
5. **72% of passengers embarked at Southampton**, making it by far the largest boarding port.
6. **Median passenger age was ~28 years**, with ages ranging from infants to 80.
7. **Fare and survival show a positive correlation** — higher fare payers had better survival odds.
8. **Passengers with 1-2 siblings/spouses aboard had slightly better survival** than those traveling alone.
9. **Female 1st class passengers had the highest survival rate (~97%)**.
10. **Children (age < 12) had relatively higher survival rates** compared to the overall average.

## 10. Conclusion

The Titanic dataset reveals strong correlations between survival and socioeconomic status, gender, and embarkation port. 
Wealth (proxied by passenger class and fare) and gender were the strongest predictors of survival. 
These findings align with historical accounts of the disaster and demonstrate the power of EDA in quickly surfacing meaningful patterns from raw data.
